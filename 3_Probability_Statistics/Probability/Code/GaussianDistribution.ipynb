## Simple example

```python
import numpy as np

heights = [48.6, 51.2, 59.1, 49.7, 58.2, 66.8, 63.1, 56.4, 54.5, 58.1]
heights = np.array(heights)
print(heights)
```

```python
mean = np.mean(heights)
var  = np.var(heights)
dev  = np.sqrt(var)

print(mean)
print(var)
print(dev)
```

```python
from scipy import stats

y_data = stats.norm.pdf(heights, mean, dev)
print(y_data)
```

```python
import math

def gaussian_function(data, mean, var):
    return (1.0/np.sqrt(2*math.pi*var)) * np.exp(-(data-mean)**2 / (2*var))

output = gaussian_function(heights, mean, var)
print(output)
```

```python
gaussian_function(np.array([35.0]), mean, var)
```

## Standard normal distribution

```python
import numpy as np
from scipy import stats

data = [1.4, 1.0, 1.3, 1.9, 2.0, 1.8]
data = np.array(data)
print(data)

mean = np.mean(data)
var  = np.var(data)

print(mean)
print(var)
```

```python
data = (data - mean) / var
print(data)
```

```python
y_data = stats.norm.pdf(data, 0, 1)
print(y_data)
```

```python
import numpy as np
import matplotlib.pyplot as plt

x_data = np.arange(0, 7, 0.1)
```

```python
import math

def gaussian_function(data, mean, var):
    return (1.0/np.sqrt(2*math.pi*var)) * np.exp(-(data-mean)**2 / (2*var))
```

```python
mean1 = 1.5666666666666667
var1  = 0.3288888888888889
y_data1 = gaussian_function(x_data, mean1, var1)
```

```python
mean2 = 3.733333333333333
var2  = 0.2722222222222222
y_data2 = gaussian_function(x_data, mean2, var2)
```

```python
plt.plot(x_data, y_data1)
plt.plot(x_data, y_data2)
```
